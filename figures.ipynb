```python
path = "./results"
```

```python
models = {}
```

```python
cls = '.ipynb_checkpoints'
```

```python
not cls.startswith('.')
```

```python
for cls in os.listdir(path):
    if (not cls.startswith('.')) and ("log_history" in cls):
        data = [values.get('loss') for values in loadjson(os.path.join(path, cls))], [values.get('epoch') for values in loadjson(os.path.join(path, cls))]
        models[cls.split("_")[0]] = data
models   
```

```python
def visualize(train_loss, validation_loss, title, img_name, epoch_number, status= "Loss"):

    fig, ax1 = plt.subplots(figsize=(10, 6), dpi=500)

    plt.title(title)

    color = 'tab:purple'
    plt.plot(epoch_number, train_loss, color=color)

    color = 'tab:blue'
    #x_axis = list(range(0, epoch_number, 3))
    plt.plot(epoch_number, validation_loss, color=color)

    class_names = ["mistral-7b-bnb-4bit", "llama-2-7b-hf"]

    plt.xlabel("Epoch")
    plt.ylabel(status + " Value")

    plt.legend(class_names, loc=1)
    plt.show()

    fig.savefig(img_name, dpi=500)
```

```python
visualize(mistral_loss, lama_loss, 'LLM Loss Comparison', 'test.jpg', mistral_epoch, status= "Loss")
```
