```python
import csv
pp = f"./2_gen_mbj_bandgap/gemma-7b-bnb-4bit_generated_samples_updated_temp__.csv"     
df1 = pd.read_csv(pp)
print(len(df1))

pp = f"./2_gen_mbj_bandgap/gemma-7b-bnb-4bit_generated_samples_updated_temp___.csv"     
df2 = pd.read_csv(pp)
print(len(df2))
```

```python
for jdx, nname in df1.iterrows():
    if jdx < 885:
        continue
    try:
        df1.loc[jdx, 'gen_material_str'] = df2.loc[jdx,'gen_material_str']
        df1.loc[jdx, 'gen_material_cif'] = df2.loc[jdx,'gen_material_cif']
        df1.loc[jdx, 'orj_material_cif'] =df2.loc[jdx,'orj_material_cif']
        df1.loc[jdx, 'rms_dist'] = df2.loc[jdx,'rms_dist']
        df1.loc[jdx, 'orj_prop_val'] = df2.loc[jdx,'orj_prop_val']
        df1.loc[jdx, 'out_data_pred'] = df2.loc[jdx,'out_data_pred']
        df1.loc[jdx, 'out_data_tar'] = df2.loc[jdx,'out_data_tar']
    except: 
        pass

df1.to_csv(f"./2_gen_mbj_bandgap/gemma-7b-bnb-4bit_generated_samples_updated.csv", index=False)
```
